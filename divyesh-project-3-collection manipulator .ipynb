```python
# ==========================================
#       STUDENT DATA ORGANIZER
# ==========================================

students = []
subjects_offered = set()

print("==========================================")
print("      WELCOME TO STUDENT DATA ORGANIZER")
print("==========================================")


# ==========================================
# 1. ADD STUDENT
# ==========================================

print("\n========== ADD STUDENTS ==========")

student_info = ("S101", "2005-05-15")

student = {
    "info": student_info,
    "name": "Rahul",
    "age": 18,
    "grade": "12th",
    "subjects": ["Python", "Maths", "English"]
}

students.append(student)

for subject in student["subjects"]:
    subjects_offered.add(subject)

print("Student added successfully!")



# Add second student

student_info = ("S102", "2006-08-20")

student = {
    "info": student_info,
    "name": "Aman",
    "age": 17,
    "grade": "11th",
    "subjects": ["Python", "Science", "Maths"]
}

students.append(student)

for subject in student["subjects"]:
    subjects_offered.add(subject)

print("Second student added successfully!")



# ==========================================
# 2. DISPLAY ALL STUDENTS
# ==========================================

print("\n========== ALL STUDENTS ==========")

if len(students) == 0:

    print("No students available.")

else:

    for student in students:

        print("\n------------------------------")

        print(f"Student ID: {student['info'][0]}")
        print(f"Name: {student['name']}")
        print(f"Age: {student['age']}")
        print(f"Grade: {student['grade']}")
        print(f"DOB: {student['info'][1]}")

        print("Subjects: {}".format(
            ", ".join(student["subjects"])
        ))



# ==========================================
# 3. UPDATE STUDENT INFORMATION
# ==========================================

print("\n========== UPDATE STUDENT ==========")

student_id = "S101"

for student in students:

    if student["info"][0] == student_id:

        # Update Age
        student["age"] = 19

        # Update Grade
        student["grade"] = "College"

        # Update Subjects
        new_subjects = ["Python", "Data Analysis", "Maths"]
        student["subjects"] = new_subjects

        print("Student information updated successfully!")

        break

else:

    print("Student not found!")



# ==========================================
# 4. DISPLAY UPDATED STUDENT
# ==========================================

print("\n========== UPDATED STUDENT ==========")

for student in students:

    if student["info"][0] == "S101":

        print("\n------------------------------")

        print(f"Student ID: {student['info'][0]}")
        print(f"Name: {student['name']}")
        print(f"Age: {student['age']}")
        print(f"Grade: {student['grade']}")
        print(f"DOB: {student['info'][1]}")

        print("Subjects: {}".format(
            ", ".join(student["subjects"])
        ))

        break



# ==========================================
# 5. DISPLAY SUBJECTS OFFERED
# ==========================================

print("\n========== SUBJECTS OFFERED ==========")

subjects_offered.clear()

for student in students:

    for subject in student["subjects"]:

        subjects_offered.add(subject)

for subject in sorted(subjects_offered):

    print("-", subject)



# ==========================================
# 6. DELETE STUDENT
# ==========================================

print("\n========== DELETE STUDENT ==========")

student_id = "S102"

for i in range(len(students)):

    if students[i]["info"][0] == student_id:

        del students[i]

        print("Student deleted successfully!")

        break

else:

    print("Student not found!")



# ==========================================
# 7. FINAL STUDENT DATA
# ==========================================

print("\n========== FINAL STUDENT DATA ==========")

for student in students:

    print("\n------------------------------")

    print(f"Student ID: {student['info'][0]}")
    print(f"Name: {student['name']}")
    print(f"Age: {student['age']}")
    print(f"Grade: {student['grade']}")
    print(f"DOB: {student['info'][1]}")
    print("Subjects:", ", ".join(student["subjects"]))


print("\n==========================================")
print("     STUDENT DATA ORGANIZER COMPLETED")
print("==========================================")
```
